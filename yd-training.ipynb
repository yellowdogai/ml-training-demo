# Train a model on a GPU, in whichever region has one

You'll train a small model on a cloud GPU. [YellowDog](https://yellowdog.ai) finds the
GPU by trying AWS regions in order until one has capacity. Then you'll see how good the
model is, and use it from Hugging Face.

### How to use this notebook

- Click a grey code cell, then press **Shift+Enter** to run it. Go **top to bottom**.
- Each step says what to do. Anything that costs money or changes your accounts needs
  a **button click**: running a cell on its own never does.
- A cell shows **[\*]** on its left while it's busy. Wait for it to finish.

Before you start, fill in `.env` in the project folder (see the README's step 1).

```python
import yd_training_ui as ui
```

## Step 1: Check

Checks your settings and sign-ins. **If anything shows ❌, follow the fix shown next to it,
then run this cell again.** Go on when everything is ✅.

```python
ui.check()
```

## Step 2: Set up

Looks up each region in `yd-training.toml` and shows what will be set up in AWS and
YellowDog, with anything to check marked in yellow. **Read it, then click Apply
configuration.** You only need to do this once, or again after changing `yd-training.toml`.

```python
ui.set_up()
```

## Step 3: Train

**Choose the settings (the defaults are fine), then click Start training.** The view
updates every 10 seconds. Zones turn red as YellowDog finds them full, and green where
the GPU starts. It takes about 10 minutes, and it's finished when it says **COMPLETED**.

```python
ui.start_training()
```

## Step 4: Results

**Run this cell once step 3 says COMPLETED.** It shows how good the model is and where it
trained. It then downloads the published model, uses it on data it hasn't seen, and
charts the predictions. It also works on its own, to show the latest published model.

```python
ui.results()
```

## Optional: Train several variants at once

**Tick the variants, then click Start sweep.** Each trains on its own GPU, possibly in
different regions, and the results are ranked. Each variant publishes to its own branch,
so the model from step 3 isn't replaced.

```python
ui.sweep()
```
